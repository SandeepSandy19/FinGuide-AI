```python
import random

for i in random.sample(range(len(finqa_train)), 3):
    qa = finqa_train[i]["qa"]

    print("=" * 80)
    print("Question:", qa["question"])
    print("Answer:", qa["answer"])
    print("Executable Answer:", qa.get("exe_ans"))
    print("Program:", qa.get("program"))
```

```python
import re
from collections import Counter

def classify_answer(answer):
    answer = str(answer).strip()

    if "%" in answer:
        return "percentage"

    if re.fullmatch(r"-?\d+", answer.replace(",", "")):
        return "integer"

    if re.fullmatch(r"-?\d+\.\d+", answer.replace(",", "")):
        return "decimal"

    return "other"


answer_types = Counter(
    classify_answer(item["qa"]["answer"])
    for item in finqa_train
)

answer_types
```

```python
for item in finqa_train:
    answer = item["qa"]["answer"]

    if classify_answer(answer) == "other":
        print("Question:", item["qa"]["question"])
        print("Answer:", answer)
        print("Program:", item["qa"].get("program"))
        print("-" * 80)
```

```python
count = 0

for item in finqa_train:
    answer = item["qa"]["answer"]

    if classify_answer(answer) == "other":
        print("Question:", item["qa"]["question"])
        print("Answer:", answer)
        print("Program:", item["qa"].get("program"))
        print("-" * 80)

        count += 1
        if count == 10:
            break
```

```python
empty_answers = [
    item for item in finqa_train
    if not str(item["qa"]["answer"]).strip()
]

print("Total training examples:", len(finqa_train))
print("Empty answers:", len(empty_answers))
print(
    "Empty answer percentage:",
    len(empty_answers) / len(finqa_train) * 100
)
```

```python
empty_answers = [
    item for item in finqa_test
    if not str(item["qa"]["answer"]).strip()
]

print("Total testing examples:", len(finqa_test))
print("Empty answers:", len(empty_answers))
print(
    "Empty answer percentage:",
    len(empty_answers) / len(finqa_test) * 100)
```
